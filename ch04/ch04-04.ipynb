## Animation

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import j1
from matplotlib.colors import LinearSegmentedColormap, PowerNorm, LightSource
import ipywidgets as widgets
from IPython.display import display

# --- 1. Global Setup ---
plot_params = {
    'font.size': 10,
    'font.family': 'sans-serif',
    'axes.linewidth': 1.0,
    'lines.linewidth': 2.0,
}
plt.rcParams.update(plot_params)

# Constants
L1 = 2 * 10**6      # Distance to screen (microns)
X_RANGE = 1e5       # microns
NUM_POINTS = 512    # Resolution

# Data Generation (Global Grid)
x1 = np.linspace(-X_RANGE, X_RANGE, num=NUM_POINTS)
y1 = x1
X1, Y1 = np.meshgrid(x1, y1)

# Pre-calculate Radial Distance on Screen (Q)
# This is the physical distance from center of screen
Q_dist = np.sqrt(X1**2 + Y1**2)

# --- 2. Helper: Wavelength to RGB ---
def wavelength_to_rgb(wavelength):
    wl = float(wavelength)
    if 380 <= wl <= 440: R,G,B = -(wl - 440)/(60), 0.0, 1.0
    elif 440 < wl <= 490: R,G,B = 0.0, (wl - 440)/(50), 1.0
    elif 490 < wl <= 510: R,G,B = 0.0, 1.0, -(wl - 510)/(20)
    elif 510 < wl <= 580: R,G,B = (wl - 510)/(70), 1.0, 0.0
    elif 580 < wl <= 645: R,G,B = 1.0, -(wl - 645)/(65), 0.0
    elif 645 < wl <= 750: R,G,B = 1.0, 0.0, 0.0
    else: R,G,B = 0.0, 0.0, 0.0
    return (R, G, B)

# --- 3. Figure Initialization (Side-by-Side) ---
fig = plt.figure(figsize=(12, 4))
fig.canvas.header_visible = False
fig.canvas.footer_visible = False

# Layout: 1 Row, 3 Columns
ax1 = fig.add_subplot(1, 3, 1)
ax2 = fig.add_subplot(1, 3, 2, projection='3d')
ax3 = fig.add_subplot(1, 3, 3)

# Initial Calculation (Airy Disk)
def calc_intensity(radius_um, wavelength_um):
    # k * a * sin(theta) approx k * a * (Q / L)
    # Argument u = (2*pi/lambda) * a * (Q/L)
    k = 2 * np.pi / wavelength_um
    u = k * radius_um * (Q_dist / L1)
    
    # Airy Pattern: (2*J1(u)/u)^2
    # Handle u=0 limit where result is 1
    with np.errstate(divide='ignore', invalid='ignore'):
        val = (2 * j1(u) / u) ** 2
    
    return np.nan_to_num(val, nan=1.0)

I_init = calc_intensity(40.0, 0.632)
rgb_init = wavelength_to_rgb(632)

# --- Plot Artists ---

# 1. Line Plot (Cross Section)
mid_idx = NUM_POINTS // 2
line_plot, = ax1.plot(x1, I_init[mid_idx, :], color=rgb_init)
ax1.set_ylim(0, 1.05)
ax1.set_xlim(-X_RANGE, X_RANGE)
ax1.set_xlabel("x (microns)")
ax1.set_ylabel("Intensity")
ax1.set_title("Cross Section")
ax1.minorticks_on()

# 2. 3D Plot (Log-Scale Height)
down = 4 # Downsample for speed (512/4 = 128x128 grid)
X_3d = X1[::down, ::down]
Y_3d = Y1[::down, ::down]
surf_plot = [None] 

def draw_3d_surface(Z_data, color_rgb):
    if surf_plot[0]: surf_plot[0].remove()
    
    Z_sub = Z_data[::down, ::down]
    
    # VISUAL ENHANCEMENT: Apply power law to height (Gamma Correction)
    # This acts like a log-scale, lifting the small rings so they are visible 3D geometry
    Z_visual = np.power(Z_sub, 0.3) 
    
    # Custom cmap scaling from black -> color -> white
    custom_cmap = LinearSegmentedColormap.from_list("laser_3d", [(0,0,0), color_rgb, (1,1,1)])
    
    surf_plot[0] = ax2.plot_surface(
        X_3d, Y_3d, Z_visual,
        rstride=1, cstride=1,
        cmap=custom_cmap,
        linewidth=0,
        antialiased=False,
        shade=False # Let the colormap handle the depth
    )

draw_3d_surface(I_init, rgb_init)
ax2.set_zlim(0, 1)
ax2.set_title("3D Profile (Log Height)")
ax2.set_xticks([]); ax2.set_yticks([]); ax2.set_zticks([])
ax2.view_init(elev=45, azim=-45)
ax2.set_facecolor('white')

# 3. Density Plot
cmap_laser = LinearSegmentedColormap.from_list("laser", [(0,0,0), rgb_init])
img_plot = ax3.imshow(I_init, extent=[-X_RANGE, X_RANGE, -X_RANGE, X_RANGE],
                      cmap=cmap_laser, norm=PowerNorm(gamma=0.4, vmin=0, vmax=1),
                      aspect='equal')
ax3.set_xticks([])
ax3.set_yticks([])
ax3.set_title("Diffraction Pattern")

plt.tight_layout()

# --- 4. Logic & Controls ---
def update_simulation(radius, wavelength):
    # Physics
    new_I = calc_intensity(radius, wavelength)
    new_color = wavelength_to_rgb(wavelength * 1000)
    
    # Update Line
    line_plot.set_ydata(new_I[mid_idx, :])
    line_plot.set_color(new_color)
    
    # Update Density
    img_plot.set_data(new_I)
    new_cmap = LinearSegmentedColormap.from_list("laser", [(0,0,0), new_color])
    img_plot.set_cmap(new_cmap)
    
    # Update 3D
    draw_3d_surface(new_I, new_color)
    
    fig.canvas.draw_idle()

def on_change(change):
    update_simulation(s_r.value, s_w.value)

s_r = widgets.FloatSlider(value=40, min=10, max=100, step=1, description='Radius (um)')
s_w = widgets.FloatSlider(value=0.632, min=0.400, max=0.700, step=0.001, description='Lambda')

s_r.observe(on_change, names='value')
s_w.observe(on_change, names='value')

display(widgets.HBox([s_r, s_w]))
```
